1.Sorteie 5 amostras de 5 % cada uma; registre médias de volume_hl. 

```python
import pandas as pd 
df = pd.read_excel("vinhos_exercicio.xlsx")
amostra_medias =[]

for i in range(5):
    amostra= df.sample(frac=0.05, random_state=i)
    media = amostra['volume_hl'].mean()
    amostra_medias.append(media)
    print(f"Amostra {i+1} - Média volume_hl: {media}")
```

2.Compare médias de amostras e a média da população.

```python
media_populacao = df["volume_hl"].mean()
print(f"\n media da populacao: {media_populacao}")
 
for i, media in enumerate(amostra_medias):
    print(f"\n Amostra {i+1}: {media}")
```
