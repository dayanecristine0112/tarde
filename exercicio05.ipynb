Crie um dicionário com 3 produtos e seus preços. Mostre o preço do produto 1.

```python
produto = [{"nome":"giz","preco":5},
           {"nome":"cola","preco":6},
           {"nome":"lapis","preco":4}]

print(produto[1]["preco"])
```
