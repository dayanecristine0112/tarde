.Crie uma função chamada par_ou_impar que receba um número como parâmetro e retorne "par" se ele for par e "ímpar" se for ímpar.


```python
def par_ou_impar(numero):
    if numero%2==0:
        print(f"o numero{numero} e par")
    else:
        print(f"o numero {numero} e impar")

par_ou_impar(2)
```

Crie uma função que converta uma temperatura de Fahrenheit para Celsius e para Kelvin e print os dois.

```python
def converta(numero):
    celsius= (numero - 32) * 5/9
    kelvin = (numero - 32) * 5/9 + 273.15
    print(f"a temperatura {numero}f convertido em celsius e {celsius} e em kelvin {kelvin}")

converta(27)
```
