1.Crie uma função contar_pares(*n) que conte e retorne quantos números pares existem nos parâmetros passados.

```python
def contar_pares(*n):
    contador = 0

    for num in n:
        if num%2 ==0:
            contador+=1
    print(f"dentro dos parametros possui {contador} numeros pares")

contar_pares(1,2,3,4,5,6)
```

Crie uma função cubica(*nums) que devolve uma lista com o cubo de cada número recebido. Use for.

```python
def cubica(*nums):
    lista=[]
    for num in nums:
        lista.append(num**3)
    print(lista)

cubica(2,5,8)
```

.Crie uma função lista_compras(**produtosValor) que receba uma lista de compras (produtos e valor) e imprima-os de forma formatada junto com o valorr total da compra.

```python
def lista_compras(**produtosValor):
    total = 0
    for produto, valor in produtosValor.items():
        print(f"{produto}: R${valor}")
        total+=valor
    
    print(f"total da compra deu R${total}")
lista_compras(arroz=20,feijao=35.56,banana=9.89)    
    
```
