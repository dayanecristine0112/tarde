1.Construa uma tabela contendo frequência absoluta, relativa (%) e acumulada da coluna grape_type.

```python
import pandas as pd
df =pd.read_excel("vinhos_exercicio.xlsx")
fa =df["grape_type"].value_counts()
fr =df["grape_type"].value_counts(normalize=True)*100
fac =fa.cumsum()
tabela =pd.DataFrame({
    "frequencia absoluta":fa, 
    "frequencia relativa":fr,
    "frequencia acumulada":fac
}) 
print(tabela)
```

2.Use pd.qcut para dividir value_usd em 4 classes iguais (quartis) e gere uma tabela com frequência absoluta e relativa de cada quartil. Use o parâmetro q=4.


```python
import pandas as pd

df["quartil"]=pd .qcut(df["value_usd"],q=4)
fa_q = df["quartil"].value_counts().sort_index()
tabela_quartil =pd.DataFrame({
    "frequencia absoluta":fa_q,
    "frequencia relativa":fa_q,
})
print(tabela_quartil)
```

3.Construa uma tabela contendo frequência absoluta, relativa (%) e acumulada da coluna year, usando as seguintes faixas (2014-2018, 2019-2022, 2023-2027).

```python
import pandas as pd

bins = [2014,2018,2022,2027]
labels=["2014-2018","2019-2022","2023-2027"]
df["faixa_ano"]= pd.cut(df["year"],bins=bins,labels=labels)
fa_year =df["faixa_ano"].value_counts().sort_index() 
fr_year =df["faixa_ano"].value_counts(normalize=True).sort_index() 
fac_year =fa_year.cumsum()
tabela_year = pd.DataFrame({
    "frequencia absoluta":fa_year, 
    "frequencia relativa":fr_year,
    "frequencia acumulada":fac_year
}) 
print(tabela_year)
```
